# Strange Places v5.1: Exploring Mysterious Phenomena

341,256 real-world mysterious phenomena from NASA, NOAA, USGS, OpenStreetMap, and the Megalithic Portal.

```python
import json
import pandas as pd
from collections import Counter

# Load the dataset
with open('all_phenomena_unified_v5.1_no_waterfalls.json') as f:
    data = json.load(f)

print(f'Total records: {len(data):,}')
```

```python
# Convert to DataFrame for analysis
df = pd.DataFrame(data)
df.head()
```

```python
# Category distribution
category_counts = df['category'].value_counts()
print('Category Distribution:')
print('=' * 40)
for cat, count in category_counts.items():
    pct = count / len(df) * 100
    print(f'{cat}: {count:,} ({pct:.1f}%)')
```

```python
# Check coordinate coverage
valid_coords = df[(df['latitude'].notna()) & (df['longitude'].notna())]
print(f'Records with valid coordinates: {len(valid_coords):,} ({len(valid_coords)/len(df)*100:.1f}%)')
```

```python
# Sample records from each category
for cat in df['category'].unique()[:5]:
    sample = df[df['category'] == cat].iloc[0]
    print(f"\n{cat}:")
    print(f"  Name: {sample.get('name', 'N/A')}")
    print(f"  Location: ({sample['latitude']:.4f}, {sample['longitude']:.4f})")
```
